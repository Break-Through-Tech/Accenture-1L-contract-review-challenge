# Milestone 2 — Task 4: Error Analysis

False negatives and false positives for multi-label contract clause prediction. Use `task4_inputs.json` exported by our shared Colab. Run this notebook after uploading the package to GitHub and cloning the PR branch, or open it from the repository.


## Load model predictions


```python
from pathlib import Path
import json
from evaluation.task4_analysis import validate_bundle, write_results

INPUT_PATH = Path("task4_inputs.json")
OUTPUT_DIR = Path("task4_results")

with INPUT_PATH.open(encoding="utf-8") as file:
    bundle = json.load(file)
labels, chunks, y_true, models = validate_bundle(bundle)
print("Rows:", len(chunks), "| Categories:", len(labels), "| Models:", list(models))
```

## Calculate false positives and false negatives


```python
reports = write_results(bundle, OUTPUT_DIR, sample_per_category=5)
category_report = reports["per_category"]
review_samples = reports["review_samples"]
pattern_report = reports["error_patterns"]
model_report = reports["model_summary"]
display(model_report)
```

## False negatives: missed labels


```python
display(category_report.sort_values(["model", "fn"], ascending=[True, False]).head(20))
display(review_samples.loc[review_samples.error_type == "FN"].head(15))
```

## False positives: extra labels


```python
display(category_report.sort_values(["model", "fp"], ascending=[True, False]).head(20))
display(review_samples.loc[review_samples.error_type == "FP"].head(15))
```

## Check common text patterns and save CSVs


```python
display(pattern_report.sort_values(["model", "matching_errors"], ascending=[True, False]).head(30))
print("Saved reports in:", OUTPUT_DIR.resolve())
```
